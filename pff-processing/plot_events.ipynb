```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
data=np.load("crab_2025-10-17_data.npy")

matrix = data[5].reshape((32, 32))
masked = np.ma.masked_where(matrix == 0, matrix)

# Plasma-Cmap kopieren und Maskenfarbe auf Weiß setzen
cmap = plt.cm.plasma.copy()
cmap.set_bad(color='white')

plt.imshow(masked, cmap=cmap, interpolation='nearest')
plt.colorbar(label='Signal (p.e.)')
plt.show()
plt.close()
```
